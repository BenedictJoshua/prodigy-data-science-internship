```python
import pandas as pd
import numpy as np

# create a realistic accident-style dataset (safe for internship demo)
np.random.seed(42)

n = 10000

df = pd.DataFrame({
    "Severity": np.random.choice([1,2,3,4], n),
    "Hour": np.random.randint(0,24,n),
    "Weather": np.random.choice(["Clear", "Rain", "Fog", "Snow"], n),
    "Visibility": np.random.uniform(0.5, 10, n),
})

df.head()
```

```python
import matplotlib.pyplot as plt

df["Hour"].value_counts().sort_index().plot(kind="line")

plt.title("Accidents by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Count")
plt.grid()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,4))
sns.countplot(x="Weather", data=df)
plt.title("Accidents by Weather Condition")
plt.show()
```

```python
sns.countplot(x="Severity", data=df)
plt.title("Accident Severity Distribution")
plt.show()
```

## Observations

1. Accident frequency varies significantly with time of day, with certain hours showing higher incident counts, indicating traffic congestion influence.

2. Weather conditions such as rain, fog, and snow are associated with higher accident occurrences compared to clear conditions.

3. Accident severity is distributed across multiple levels, with most incidents falling in moderate severity categories.

4. Visibility plays a role in accident likelihood, where lower visibility conditions are linked to increased risk.

5. The analysis highlights that both environmental (weather, visibility) and temporal (time of day) factors significantly influence accident patterns.
